```python
#audit mssing value

missing_count = df.isnull().sum()
null_count = df.isna().sum()
missing_pct = (df.isna().mean() + 100).round(2)
report = pd.DataFrame({
    "null" : null_count,
    "missing" : missing_count,
    "persentase" : missing_pct
}).sort_values(by="missing", ascending=False)
display(report)
```

```python
df_clean = df.copy()
print("duplikasi : ", df_clean.duplicated().sum())
df_clean = df_clean.drop_duplicates()

for col in["type", "rating", "country"]:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].astype("string").str.strip()

for col in ["director", "cast", "country", "rating"]:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].fillna("Unknown")

display(df_clean)
```

```python
df_clean["data_added"] = pd.to_datetime(df_clean["date_added"], errors="coerce")

print(df_clean["data_added"].dtype)
print(df_clean["data_added"].isnull().sum())
display(df_clean["data_added"])
```

```python
df_clean["duration_value"] = pd.to_numeric(
    df_clean["duration"].astype("string").str.extract(r"(\d+)")[0],errors="coerce"
)
print(df_clean[["type", "duration", "duration_value"]].head(10))
```

```python
movies = df_clean[df_clean["type"] == "Movie"].copy()

movies["duration_min"] = pd.to_numeric(
    movies["duration"].astype("string")
    .str.extract(r"(\d+)")[0], errors="coerce"
)

q1 = movies["duration_min"].quantile(.25)
q3 = movies["duration_min"].quantile(.75)
iqr = q3 - q1

low, high = q1 - 1.5*iqr, q3 + 1.5*iqr

outliers = movies[~movies["duration_min"].between(low, high)]

display(outliers)
```
